```python
import mercury as mr
import os
print('KERNEL_PID', os.getpid())
```

```python
chat = mr.Chat()
```

```python
prompt = mr.ChatInput(placeholder='Security test input')
```

```python
if prompt.value:
    if prompt.value.startswith('MD:'):
        rendered = mr.Markdown(prompt.value[3:])
    elif prompt.value.startswith('STREAM:'):
        message = mr.Message()
        chat.add(message)
        for chunk in prompt.value[7:].split('SPLIT'):
            message.append_markdown(chunk)
    else:
        chat.add(mr.Message(prompt.value, role='user'))
```
