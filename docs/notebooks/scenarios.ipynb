# What-if forecast

Save **Baseline** with 5% growth and $50 price. Try **High growth** (15%, $50) and **Low price** (5%, $35), save each, then compare values, tables, and charts together. Saved scenarios remain in this browser after refresh.

```python
import mercury as mr
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
growth = mr.Slider(value=5, min=0, max=30, label="Monthly growth (%)", key="growth")
```

```python
price = mr.NumberInput(value=50, min=10, max=100, label="Price ($)", key="price")
```

```python
presets = mr.Examples({
    "Baseline": {"growth": 5, "price": 50},
    "High growth": {"growth": 15, "price": 50},
    "Low price": {"growth": 5, "price": 35},
}, key="forecast-presets")
```

```python
months = list(range(1, 13))
units = [1000 * (1 + growth.value / 100) ** (month - 1) for month in months]
forecast = pd.DataFrame({
    "Month": months,
    "Units": [round(count) for count in units],
    "Revenue": [round(count * price.value, 2) for count in units],
    "Profit": [round(count * (price.value - 25), 2) for count in units],
})
revenue = forecast["Revenue"].sum()
profit = forecast["Profit"].sum()
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(months, forecast["Revenue"], marker="o", label="Revenue")
ax.plot(months, forecast["Profit"], marker="o", label="Profit")
ax.set(xlabel="Month", ylabel="USD", title="Monthly forecast")
ax.legend()
fig.tight_layout()
plt.close(fig)
```

```python
mr.Indicator([
    mr.Indicator(value=f"${revenue:,.0f}", label="Annual revenue"),
    mr.Indicator(value=f"${profit:,.0f}", label="Annual profit"),
])
```

```python
display(fig)
```

```python
mr.Table(forecast)
```

```python
scenarios = mr.Scenarios(
    inputs={"Growth (%)": growth, "Price ($)": price},
    outputs={
        "Annual revenue": f"${revenue:,.0f}",
        "Annual profit": f"${profit:,.0f}",
        "Monthly forecast": forecast,
        "Forecast chart": fig,
    },
    key="forecast-scenarios",
)
```
