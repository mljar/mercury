# Camera showcase

Allow camera access to see the live preview, then take a photo or record a short video.

```python
import mercury as mr
```

```python
capture_mode = mr.Select(
    label="Capture mode",
    value="Photo",
    choices=["Photo", "Video"],
    key="camera-mode",
)
```

```python
camera = mr.Camera(
    mode=capture_mode.value.lower(),
    label="Live camera",
    facing_mode="environment",
    max_duration=10,
    key=f"showcase-{capture_mode.value.lower()}",
)
```

```python
if camera.value is None:
    print("Take a photo or record a video to send it to Python.")
else:
    print(f"Captured: {camera.filename}")
    print(f"MIME type: {camera.mime_type}")
    print(f"Size: {camera.size:,} bytes")
    if camera.mode == "video":
        print(f"Duration: {camera.duration:.1f} seconds")
```

```python
if camera.value is not None and camera.mode == "photo":
    from IPython.display import Image, display

    display(Image(data=camera.value))
```
