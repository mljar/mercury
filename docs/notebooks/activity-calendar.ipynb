# GitHub outage activity calendar

This example displays prepared daily outage data. Missing dates are rendered as inactive days.

```python
import pandas as pd
import mercury as mr
```

```python
outages = pd.DataFrame({
    "date": [
        "2026-08-02",
        "2026-08-05",
        "2026-08-11",
        "2026-08-20",
        "2026-08-21",
        "2026-08-23",
        "2026-08-29",
    ],
    "outage_hours": [0.2, 1.4, 0, 0.5, 2.1, 4.8, 1.0],
})
```

```python
mr.ActivityCalendar(
    outages,
    date="date",
    value="outage_hours",
    title="GitHub outages",
    unit="hours",
    color="red",
    start_date="2026-08-01",
    end_date="2026-08-31",
)
```
