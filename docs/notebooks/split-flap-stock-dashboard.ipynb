# Market board

A split-flap stock dashboard using deterministic demo data. Click **Next market tick** to animate the changed characters.

```python
import pandas as pd
import mercury as mr
```

```python
# Static demo data keeps this example fast, reproducible, and network-free.
previous_close = {
    "AAPL": 231.18,
    "MSFT": 417.52,
    "NVDA": 139.91,
    "TSLA": 347.68,
}

snapshots = pd.DataFrame(
    [
        {"time": "09:30:00", "AAPL": 231.42, "MSFT": 417.10, "NVDA": 140.22, "TSLA": 346.95},
        {"time": "09:30:05", "AAPL": 231.57, "MSFT": 417.36, "NVDA": 140.48, "TSLA": 347.21},
        {"time": "09:30:10", "AAPL": 231.31, "MSFT": 417.84, "NVDA": 140.37, "TSLA": 348.04},
        {"time": "09:30:15", "AAPL": 231.88, "MSFT": 418.03, "NVDA": 140.91, "TSLA": 347.76},
        {"time": "09:30:20", "AAPL": 232.14, "MSFT": 417.69, "NVDA": 141.16, "TSLA": 348.53},
    ]
)

def market_board_text(snapshot):
    lines = [
        f"NASDAQ  {snapshot['time']} ET",
        "SYMBOL    PRICE   CHANGE",
    ]
    for ticker in previous_close:
        price = snapshot[ticker]
        change = (price / previous_close[ticker] - 1) * 100
        arrow = "▲" if change >= 0 else "▼"
        price_text = f"${price:,.2f}"
        move_text = f"{arrow}{abs(change):.2f}%"
        lines.append(f"{ticker:<6}{price_text:>9}  {move_text:>7}")
    return "\n".join(lines)

first = snapshots.iloc[0]
```

## NASDAQ · demo feed

```python
market_board = mr.SplitFlap(
    market_board_text(first),
    size="small",
    key="stock-market-board",
)
```

```python
next_tick = mr.Button(
    label="Next market tick",
    position="sidebar",
    key="next-market-tick",
)
```

```python
# This cell is below the button, so Mercury reruns it after every click.
# Assigning to .value updates the already-displayed board in place.
snapshot = snapshots.iloc[next_tick.n_clicks % len(snapshots)]

market_board.value = market_board_text(snapshot)
```
