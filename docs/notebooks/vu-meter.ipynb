# Analog instrument panel

Retro VU meters for live application and model telemetry. Click **Next reading** to move the needles.

```python
import mercury as mr
```

```python
# Deterministic readings keep the showcase reproducible and network-free.
readings = [
    {"cpu": 28, "confidence": 0.62, "anomaly": 0.12},
    {"cpu": 47, "confidence": 0.74, "anomaly": 0.27},
    {"cpu": 73, "confidence": 0.82, "anomaly": 0.44},
    {"cpu": 91, "confidence": 0.93, "anomaly": 0.86},
    {"cpu": 64, "confidence": 0.88, "anomaly": 0.31},
]
first = readings[0]
```

## Live telemetry

```python
cpu_col, confidence_col, anomaly_col = mr.Columns(
    3,
    min_width="240px",
    gap="18px",
    border="",
    key="vu-meter-panel",
)

with cpu_col:
    cpu_meter = mr.VUMeter(
        first["cpu"],
        min=0,
        max=100,
        label=f"CPU LOAD · {first['cpu']}%",
        zones=[60, 85],
        higher_is_better=False,
        key="showcase-cpu",
    )

with confidence_col:
    confidence_meter = mr.VUMeter(
        first["confidence"],
        min=0,
        max=1,
        label="MODEL CONFIDENCE",
        zones=[0.5, 0.8],
        key="showcase-confidence",
    )

with anomaly_col:
    anomaly_meter = mr.VUMeter(
        first["anomaly"],
        min=0,
        max=1,
        label="ANOMALY SCORE",
        zones=[0.35, 0.7],
        higher_is_better=False,
        key="showcase-anomaly",
    )
```

```python
next_reading = mr.Button(
    label="Next reading",
    position="sidebar",
    key="vu-meter-next-reading",
)
```

```python
# This cell reruns after a click and updates the existing instruments.
reading = readings[next_reading.n_clicks % len(readings)]
cpu_meter.value = reading["cpu"]
cpu_meter.label = f"CPU LOAD · {reading['cpu']}%"
confidence_meter.value = reading["confidence"]
anomaly_meter.value = reading["anomaly"]
```

## Theme-aware by design

The dial uses the active Mercury theme for its casing, face, typography, needle, semantic zones, radius, and shadow. Change those values in `config.toml`; the notebook code stays the same.

## Size variants

```python
small_col, medium_col, large_col = mr.Columns(
    3, min_width="220px", gap="18px", border="", key="vu-sizes"
)

with small_col:
    mr.VUMeter(36, label="SMALL", size="small", key="vu-small")

with medium_col:
    mr.VUMeter(58, label="MEDIUM", size="medium", key="vu-medium")

with large_col:
    mr.VUMeter(
        79,
        label="LARGE",
        size="large",
        key="vu-large",
    )
```
