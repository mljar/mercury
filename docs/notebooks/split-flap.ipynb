# Split-Flap Display

Retro animated boards for KPIs, counters, rankings, scores, and dashboard headlines.

```python
import mercury as mr
```

## Numeric KPI

```python
users = mr.SplitFlap(
    value="12,482",
    key="total-users",
)
```

## Large currency display

```python
revenue = mr.SplitFlap(
    value="$42,810",
    size="large",
    key="revenue",
)
```

## Short text and compact sizing

```python
departure = mr.SplitFlap(
    value="ON TIME",
    size="small",
    key="departure",
)
```

## Immediate updates without animation

```python
rank = mr.SplitFlap(
    value="#03",
    animate=False,
    key="rank",
)
```

## Interactive score board

Move the slider to trigger the mechanical flip transition on changed digits.

```python
score = mr.Slider(
    label="Score",
    value=42,
    min=0,
    max=999,
    position="inline",
    key="score-control",
)
```

```python
scoreboard = mr.SplitFlap(
    value=f"{score.value:03}",
    size="large",
    key="scoreboard",
)
```
