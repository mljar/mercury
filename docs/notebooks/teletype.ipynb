## Incremental agent log

Click **Append next event** to add one complete event to the existing animation queue.

```python
events = [
    "[00:01] Planner selected 3 tools.\n",
    "[00:02] Reading customer_metrics.parquet...\n",
    "[00:03] Validated 18,420 records ✓\n",
    "[00:04] Forecast confidence: 92.4%\n",
    "[00:05] Agent finished successfully. 🤖\n",
]
```

```python
agent_log = mr.Teletype(
    "$ agent run forecast\n",
    speed=16,
    auto_scroll=True,
    key="agent-log",
)
```

```python
next_event = mr.Button(
    label="Append next event",
    position="inline",
    key="teletype-next-event",
)
```

```python
# This cell reruns after a click and appends only the next event.
if 1 <= next_event.n_clicks <= len(events):
    agent_log.append(events[next_event.n_clicks - 1])
```

## Provider-paced stream

Use `speed=0` when chunks already arrive at the desired pace.

```python
stream = mr.Teletype("", speed=0, cursor=True, key="stream-output")
```

```python
for chunk in ["Mercury ", "renders incoming ", "plain-text chunks immediately."]:
    stream.append(chunk)
```
