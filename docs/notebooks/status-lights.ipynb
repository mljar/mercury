# Status Lights

Animated status lamps for services, pipelines, ML workflows, and agent execution.

```python
import mercury as mr
```

## All states

```python
all_states = mr.StatusLights(
    {
        "Offline": "off",
        "Database": "ok",
        "Model": "warning",
        "Worker": "error",
        "Agent": "active",
    },
    title="System health",
    key="all-states",
)
```

## Blink and pulse variants

```python
animated = mr.StatusLights(
    {
        "API": "ok",
        "Queue": "active",
        "GPU": "warning",
        "Failed job": "error",
    },
    blink=["Failed job"],
    pulse=["Queue", "GPU"],
    title="Animated signals",
    show_group_border=False,
    key="animations",
)
```

## Vertical sidebar without title or item borders

In the Mercury app this compact group is placed in the sidebar.

```python
sidebar_lights = mr.StatusLights(
    {"Scheduler": "ok", "Agent": "active", "Alerts": "off"},
    pulse=["Agent"],
    title="Runtime",
    orientation="vertical",
    show_title=False,
    show_item_borders=False,
    position="sidebar",
    key="sidebar-runtime",
)
```

## Live pipeline updates

Move the slider to advance the pipeline. Mercury reruns the cell below it and updates the same keyed light group.

```python
pipeline_step = mr.Slider(
    label="Pipeline progress",
    value=0,
    min=0,
    max=3,
    position="inline",
    key="pipeline-progress",
)
```

```python
stages = ["Extract", "Transform", "Load"]
step = pipeline_step.value
statuses = {
    stage: (
        "ok" if index < step
        else "active" if index == step and step < len(stages)
        else "off"
    )
    for index, stage in enumerate(stages)
}
running = [stages[step]] if step < len(stages) else []

pipeline = mr.StatusLights(
    statuses,
    pulse=running,
    title=f"ETL pipeline · step {step}/3",
    key="etl-pipeline",
)
```
