# Customer examples

Choose a predefined customer in the sidebar to populate all inputs at once.

```python
import mercury as mr
```

```python
age = mr.NumberInput(
    label="Age",
    value=32,
    min=18,
    max=100,
    key="age",
)
```

```python
country = mr.Select(
    label="Country",
    value="Poland",
    choices=["Poland", "Germany", "France", "United States"],
    url_key="country",
)
```

```python
segment = mr.Select(
    label="Customer segment",
    value="Consumer",
    choices=["Consumer", "Professional", "Enterprise"],
    key="segment",
)
```

```python
monthly_spend = mr.NumberInput(
    label="Monthly spend ($)",
    value=250,
    min=0,
    max=50000,
    step=50,
    key="monthly_spend",
)
```

```python
newsletter = mr.CheckBox(
    label="Newsletter subscriber",
    value=True,
    key="newsletter",
)
```

```python
presets = mr.Examples(
    examples={
        "Young customer": {
            "age": 25,
            "country": "Poland",
            "segment": "Consumer",
            "monthly_spend": 120,
            "newsletter": True,
        },
        "International professional": {
            "age": 38,
            "country": "France",
            "segment": "Professional",
            "monthly_spend": 1450,
            "newsletter": False,
        },
        "Enterprise customer": {
            "age": 45,
            "country": "Germany",
            "segment": "Enterprise",
            "monthly_spend": 12500,
            "newsletter": True,
        },
    },
    key="customer-examples",
)
```

## Current profile

```python
summary = mr.Markdown(
    f"### {segment.value} profile\n\n"
    f"- **Age:** {age.value:.0f}\n"
    f"- **Country:** {country.value}\n"
    f"- **Monthly spend:** ${monthly_spend.value:,.0f}\n"
    f"- **Newsletter:** {'Subscribed' if newsletter.value else 'Not subscribed'}"
)
```
