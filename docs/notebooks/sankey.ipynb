# Customer funnel Sankey diagram

Ribbon width represents the number of customers moving between funnel stages.

```python
import pandas as pd
import mercury as mr

funnel = pd.DataFrame({
    "source": [
        "Website",
        "Website",
        "Signup",
        "Signup",
        "Trial",
        "Trial",
    ],
    "target": [
        "Signup",
        "Left",
        "Trial",
        "Left",
        "Paid",
        "Left",
    ],
    "value": [1000, 400, 600, 250, 180, 170],
})
```

```python
mr.Sankey(
    funnel,
    source="source",
    target="target",
    value="value",
    height=500,
    show_values=True,
    value_format=",",
    colors=["#3b82f6", "#22c55e", "#f59e0b", "#ef4444"],
)
```
