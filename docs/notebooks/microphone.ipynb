# Microphone showcase

Record a short voice note, review it in the audio player, and access the completed recording in Python.

```python
import mercury as mr
```

```python
microphone = mr.Microphone(
    label="Record a voice note",
    max_duration=15,
    key="voice-note",
)
```

```python
if microphone.value is None:
    print("Select Start recording to create an audio sample.")
else:
    print(f"Recorded: {microphone.filename}")
    print(f"MIME type: {microphone.mime_type}")
    print(f"Duration: {microphone.duration:.1f} seconds")
    print(f"Size: {microphone.size:,} bytes")
```

```python
if microphone.value is not None:
    from IPython.display import Audio, display

    display(Audio(data=microphone.value))
```
