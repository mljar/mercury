# Sales funnel

Each section represents one stage. Width represents its value, and the percentage shows conversion from the previous stage.

```python
import pandas as pd
import mercury as mr

sales = pd.DataFrame({
    "stage": ["Visitors", "Signups", "Trials", "Customers"],
    "users": [10000, 3200, 1200, 340],
})
```

```python
mr.Funnel(sales, stage="stage", value="users")
```

## Compare every stage with the first stage

```python
mr.Funnel(
    sales,
    stage="stage",
    value="users",
    height=420,
    percentage="first",
    colors=["#1d4ed8", "#2563eb", "#3b82f6", "#60a5fa"],
)
```

## Small, increasing, zero, and single-stage funnels

Positive values remain visible even after a very large drop. Increasing and zero stages are valid.

```python
mr.Funnel([("Visitors", 100000), ("Signup", 1000), ("Paid", 10)])
```

```python
mr.Funnel([("Invited", 100), ("Registered", 120), ("Paid", 20)])
```

```python
mr.Funnel([("Visitors", 1000), ("Signup", 200), ("Paid", 0)])
```

```python
mr.Funnel([("Users", 1200)])
```
